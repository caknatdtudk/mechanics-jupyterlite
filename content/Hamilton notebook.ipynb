# Analytisk mekanik
## Hamilton operator
Lagrange funktionalet defineres som forskellen i kinetisk og potentiel energi.
$$L=T-V$$

Bevægelsesligningen for systemet bliver
$$\frac{d}{dt}\frac{\partial L}{\partial \dot q}-\frac{\partial L}{\partial q}=0$$

Vi indfører den generaliserede impuls
$$p_q=\frac{\partial L}{\partial \dot q}$$
Vi definerer Hamilton operatoren
$$H=\sum_i p_iq_i-L$$
De såkaldte kanoniske bevægelsesligninger bliver
$$\frac{dq}{dt}=\frac{\partial H}{\partial p_q}$$
og
$$\frac{dp_q}{dt}=-\frac{\partial H}{\partial q}$$

I dette dokument vil vi opstille bevægelsesligninger for forskellige systemer baseret på $H$.

## Harmonisk oscillator
Kinetisk energi
$$T=\frac{1}{2}m\dot x^2$$
Potentiel energi
$$V=\frac{1}{2}k x^2$$
Generaliseret impuls
$$p_x=\frac{\partial L}{\partial \dot x}=m\dot x$$
Hamilton operator
$$H=p_x\dot x-L$$
$$H=\frac{p_x^2}{m}-\frac{p_x^2}{2m}+\frac{1}{2}k x^2$$
$$H=\frac{p_x^2}{2m}+\frac{1}{2}k x^2$$

I omskrivningen til $H$  indgår den generaliserede hastighed. 
Det er nødvendigt omskrive denne så det kun er den generaliserede impuls der indgår.

```python
import sympy as sp
t = sp.symbols('t')
x = sp.Function('x')(t)
m,k = sp.symbols('m,k')
T = sp.Rational(1,2)*m*x.diff(t)**2
V = sp.Rational(1,2)*k*x**2
L = T - V
print('Lagrange funktional')
display(L)
p_x = L.diff(x.diff(t))
print('Generaliseret impuls')
display(p_x)
H = p_x*x.diff(t) - L
print('Hamilton operator')
display(H)
print(str(H))
p_x = sp.Function('p_x')(t)
H = H.subs(x.diff(t),p_x/m)
print('Hamilton operator omskrevet til generaliseret impuls')
display(H)
eq1 = sp.Eq(x.diff(t),H.diff(p_x))
eq2 = sp.Eq(p_x.diff(t),-H.diff(x))
print('Bevægelsesligninger')
display(eq1)
display(eq2)
```

## Pendul

```python
import sympy as sp
t = sp.symbols('t')
theta = sp.Function('theta')(t)
m,r,g = sp.symbols('m,r,g')
T = sp.Rational(1,2)*m*r**2*theta.diff(t)**2
V = -m*g*r*sp.cos(theta)
L = T - V
print('Lagrange funktional')
display(L)
print('Bevægelsesligning')
display(sp.Eq(L.diff(theta.diff(t)).diff(t),L.diff(theta)))
p_theta = L.diff(theta.diff(t))
print('Generaliseret impuls')
display(p_theta)
H = p_theta*theta.diff(t) - L
print('Hamilton operator')
display(H)
#print('Som en streng')
#print(str(H))
p_theta = sp.Function('p_theta')(t)
H = H.subs(theta.diff(t),p_theta/(m*r**2))
print('Hamilton operator')
display(H)
eq1 = sp.Eq(theta.diff(t),H.diff(p_theta))
eq2 = sp.Eq(p_theta.diff(t),-H.diff(theta))
print('Bevægelsesligninger')
display(eq1)
display(eq2)
```

## Pendul med lodret bevæget ophæng

Ophængnignsunktet for pendulet tænkes at bevæge sig op og ned harmonisk som beskrevet ved funktionen $A\sin(\omega t)$.
Akserne vælges op ad og til højre.
Udtrykken for den kinetiske energi er lidt vanskelige at opstille med den generaliserede koordinat, $\theta$,
derfor regner vi første de kartesiske koordinater ud, finder hastighed ud fra disse og så opstilles udtryk for
de kinetiske (og potentielle) energier.


```python
import sympy as sp
t = sp.symbols('t')
theta = sp.Function('theta')(t)
v_theta = sp.Function('v_theta')(t)
m,l,g,A,omega = sp.symbols('m,l,g,A,omega')
x = l*sp.sin(theta)
y = A*sp.sin(omega*t)-l*sp.cos(theta)
v_x = x.diff(t)
v_y = y.diff(t)
print('x')
display(x)
print('y')
display(y)
print('v_x')
display(v_x)
print('v_y')
display(v_y)
T = sp.Rational(1,2)*m*(v_x**2+v_y**2)
V = m*g*y
print('T')
display(T)
T = T.expand()
T = T.simplify()
display(T)
print('V')
display(V)
L = T - V
print('L')
display(L)
L = L.expand()
L = L.simplify()
print('L')
display(L)
p_theta = L.diff(theta.diff(t))
print('p_theta')
display(p_theta)
H = p_theta*theta.diff(t) - L
display(H)
print('Som en streng')
print(str(H))
H = H.subs(theta.diff(t),((2*p_theta/m-2*A*l*omega*sp.sin(theta)*sp.cos(omega*t))/(2*l**2)))
print('Hamilton operator')
display(H)
```
