#### Free fall with air resistance

$$m\frac{dv}{dt}=mg-bv$$

To nondimensionalize the equation we introduce $v=k_vV$ and $t=k_tT$ where $V$ and $T$ should have dimension 1.




```python
import sympy as sp
from sympy import symbols,Function,Eq,diff
m,b,g = symbols('m,b,g',positive=True)
t = symbols('t')
v = Function('v')(t)
eq = Eq(m*diff(v,t),-b*v+m*g)
display(eq)
```

```python
kv,kt = symbols('k_v,k_t')
T = symbols('T')
V = Function('V')(T)
eq = eq.subs({v:kv*V,v.diff(t):kv/kt*V.diff(T)})
display(eq)
```

```python
eq = Eq(eq.lhs/(kv*m/kt),eq.rhs/(kv*m/kt))
display(eq.expand())
```

```python
eq = eq.subs({kt:m/b})
display(eq.expand())
```

```python
eq = eq.subs({kv:m*g/b})
display(eq.expand())
```

#### Nondimensionalization of damped oscillator (I/II)

$m\frac{d^2x}{dt^2}+b\frac{dx}{dt}+kx=0$

$m\ddot{x}+b\dot{x}+kx=0$

```python
import sympy as sp
from sympy import symbols,Function,Eq,diff
m,b,k = symbols('m,b,k',positive=True)
t = symbols('t')
x = Function('x')(t)
eq = Eq(m*diff(x,t,t)+b*x.diff(t)+k*x,0)
display(eq)
```

Insert scaling $x=k_xX$ and $t=k_tT$. 
This is accomplished by replacing 
$x$ with $k_xX$,
$\frac{dx}{dt}$ with $\frac{k_xdX}{k_tdT}$, and
$\frac{d^2x}{dt^2}$ with $\frac{k_xd^2X}{k_t^2dT^2}$.

```python
kx,kt = symbols('k_x,k_t')
T = symbols('T')
X = Function('X')(T)
eq = eq.subs({x:kx*X,x.diff(t):kx/kt*X.diff(T),x.diff(t,t):kx/kt**2*X.diff(T,T)})
display(eq)
```

Note that $k_x$ is a common factor so we eliminate it.

```python
eq = Eq(eq.lhs/kx,eq.rhs/kx)
display(eq)
```

We need to expand the equation in order to simply the expression.

```python
eq = eq.expand()
display(eq)
```

We can simplify the second order derivative by multiplying the equation by $k_t^2/m$
and expanding the expression.

```python
eq = Eq(eq.lhs*kt**2/m,eq.rhs*kt**2/m)
eq = eq.expand()
display(eq)
```

To simplify the equation we can pick $k_t=m/b$.

```python
eq = eq.subs({kt:m/b})
display(eq)
```

We see that there is a single dimensionless parameter, $km/b^2$, a dimensionless spring constant.

We now repeat the analysis with a different choice for $k_t$.

```python
import sympy as sp
from sympy import symbols,Function,Eq,diff,sqrt
m,b,k = symbols('m,b,k',positive=True)
t = symbols('t')
x = Function('x')(t)
eq = Eq(m*diff(x,t,t)+b*x.diff(t)+k*x,0)
display(eq)
```

Insert the scaling.

```python
kx,kt = symbols('k_x,k_t')
T = symbols('T')
X = Function('X')(T)
eq = eq.subs({x.diff(t,t):kx/kt**2*X.diff(T,T),x.diff(t):kx/kt*X.diff(T),x:kx*X})
display(eq)
```

Eliminate the common factor, $k_x$.

```python
eq = Eq(eq.lhs/kx,eq.rhs/kx)
eq = eq.expand()
display(eq)
```

Simplify the expression by simplifying second order term.

```python
eq = Eq(eq.lhs*kt**2/m,eq.rhs*kt**2/m)
eq = eq.expand()
display(eq)
```

We pick the scaling constant $k_t=\sqrt{m/k}$.

```python
eq = eq.subs({kt:sqrt(m/k)})
display(eq)
```

There is a single dimensionless parameter, $b/\sqrt{mk}$, a dimensionless damping coefficient.

#### Nondimensionalization of forced oscillator

```python
import sympy as sp
from sympy import symbols,Function,Eq,Derivative,diff
m,b,k,F,omega=symbols('m,b,k,F,omega',positive=True)
t = symbols('t')
x = Function('x')(t)
eq=Eq(m*sp.diff(x,t,t)+b*sp.diff(x,t)+k*x,F*sp.sin(omega*t))
display(eq)
kx,kt=symbols('k_x,k_t')
T = symbols('T')
X = Function('X')(T)
Y = Function('Y')(T)
eq=eq.subs({diff(x,t,t):kx/kt**2*diff(X,T,T),diff(x,t):kx/kt*sp.diff(X,T),x:kx*X,t:kt*T})
display(eq)
eq=Eq(kt**2/kx/m*eq.lhs,kt**2/kx/m*eq.rhs)
display(eq)
eq=eq.expand()
display(eq)
eq=eq.subs({kt:1/omega})
display(eq)
eq=eq.expand()
eq=eq.subs({kx:F/m/omega**2})
display(eq)
eq=eq.expand()
display(eq)
```
