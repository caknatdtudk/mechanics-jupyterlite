# Analytisk mekanik
## Lagrange funktional
Lagrange funktionalet defineres som forskellen i kinetisk og potentiel energi.
$$L=T-V$$

Energierne opskrives som funktioner af generaliserede koordinater, $q$ og generaliserede hastigheder, $\dot q$.

Bevægelsesligningen for systemet er
$$\frac{d}{dt}\frac{\partial L}{\partial \dot q}-\frac{\partial L}{\partial q}=0$$

Der er en bevægelsesligning per generaliseret koordinat.

I dette dokument vil vi opstille bevægelsesligninger for forskellige systemer baseret på lagrangefunktionalet
ved brug af Python pakken *sympy*.

## Harmonisk oscillator
Vi betragter en harmonisk oscillator der består af en partikel med massen $m$ der er fastgjort til en væg med en fjeder. 
Fjederen har fjederkonstanten, $k$.


Positionen af massen, $x$, (svarende til forlængelse af fjederen), er en generaliseret koordinat for systemet.

Kinetisk energi
$$T=\frac{1}{2}m\dot x^2$$
Potentiel energi
$$V=\frac{1}{2}k x^2$$
Lagrange funktional
$$L=T-V$$
$$L=\frac{1}{2}m\dot x^2-\frac{1}{2}k x^2$$

````python
import sympy as sp
t = sp.symbols('t') # tiden
x = sp.Function('x')(t) # generaliseret koordinat der er en funktion af tiden
m,k = sp.symbols('m,k')
T = sp.Rational(1,2)*m*x.diff(t)**2
V = sp.Rational(1,2)*k*x**2
L = T - V
print('Lagrange funktional')
display(L)
eq1 = sp.Eq(L.diff(x.diff(t)).diff(t)-L.diff(x),0)
print('Bevægelsesligning')
display(eq1)
````

Man kan også definere den generaliserede hastighed hvilket kan gøre koden lidt mere læselig (output er det samme).

````python
import sympy as sp
t = sp.symbols('t') # tiden
x = sp.Function('x')(t) # generaliseret koordinat der er en funktion af tiden
v = x.diff(t) # generaliseret hastighed
m,k = sp.symbols('m,k')
T = sp.Rational(1,2)*m*v**2
V = sp.Rational(1,2)*k*x**2
L = T - V
print('Lagrange funktional')
display(L)
eq1 = sp.Eq(L.diff(v).diff(t)-L.diff(x),0)
print('Bevægelsesligning')
display(eq1)
````

## Pendul
Vinklen $\theta(t)$ der er snorens vinkel med vandret, er generaliseret koordinat.

````python
import sympy as sp
t = sp.symbols('t')
theta = sp.Function('theta')(t)
omega = theta.diff(t)
m,r,g = sp.symbols('m,r,g')
T = sp.Rational(1,2)*m*r**2*omega**2
print('Kinetisk energi')
display(T)
V = -m*g*r*sp.cos(theta)
print('Potentiel energi')
display(V)
L = T - V
print('Lagrange funktional')
display(L)
eq1 = sp.Eq(L.diff(omega).diff(t)-L.diff(theta),0)
print('Bevægelsesligning')
display(eq1)
print('Bevægelsesligning (forsimplet)')
display(sp.Eq(eq1.lhs/(m*r**2),eq1.rhs/(m*r**2)).simplify()) # vi forsimpler bevægelsesligningen lidt 
````

## Pendul med lodret bevæget ophæng

Ophængningsunktet for pendulet tænkes nu at bevæge sig op og ned harmonisk som beskrevet ved en funktion, $A\sin(\omega t)$.
Positive akserne vælges op ad og til højre.
Udtrykket for den kinetiske energi er lidt mere vanskeligt at opstille med den generaliserede koordinat, $\theta$,
derfor regner vi i stedet først de kartesiske koordinater ud, finder hastighed ud fra disse og så opstilles udtryk for
kinetisk og potentiel energi.

Det er ikke altid at udtryk er simplificeret, derfor benyttes metoderne *expand* og *simplify* for at reducere et udtryks kompleksit.
```python
L = L.expand()
L = L.simplify()
```
Bemærk at begge metoder returnerer et nyt udtryk der så skal gemmes, det ændrer i på det oprindelige udtryk.

````python
import sympy as sp
t = sp.symbols('t')
theta = sp.Function('theta')(t)
v_theta = sp.Function('v_theta')(t)
m,l,g,A,omega = sp.symbols('m,l,g,A,omega')
x = l*sp.sin(theta) # x-koordinat for partiklen
y = A*sp.sin(omega*t)-l*sp.cos(theta) # # y-koordinat for partiklen
v_x = x.diff(t) # hastighed i x-retning
v_y = y.diff(t) # hastighed i y-retning
print('x koordinat')
display(x)
print('y koordinat')
display(y)
print('v_x i x-retning')
display(v_x)
print('v_y i y-retning')
display(v_y)
T = sp.Rational(1,2)*m*(v_x**2+v_y**2)
V = m*g*y
print('Kinetisk energi T')
display(T)
T = T.expand()
T = T.simplify()
print('Kinetisk energi T (expand og simplify)')
display(T)
print('Potentiel energi V')
display(V)
L = T - V
print('L')
display(L)
L = L.expand()
L = L.simplify()
print('L (expand og simplify)')
display(L)
eq = sp.Eq(L.diff(theta.diff(t),t)-L.diff(theta),0)
eq = eq.expand()
eq = eq.simplify()
print('Bevægelsesligning')
display(eq)
````

## Fjederpendul
Et simpelt pendul som tidligere men snoren er nu elastisk med fjederkonstant, $k$.

Vi beskriver bevægelsen i polære koordinater, så $r(t)$ og $\theta(t)$ er de generaliserede koordinater.

````python
import sympy as sp
t = sp.symbols('t')
r = sp.Function('r')(t)
theta = sp.Function('theta')(t)
m,k,r0,g = sp.symbols('m,k,r0,g')
T = sp.Rational(1,2)*m*(r.diff(t)**2+r**2*theta.diff(t)**2)
V = -m*g*(r-r0)*sp.cos(theta)+sp.Rational(1,2)*k*(r-r0)**2
#print('T')
#display(T)
#print('V')
#display(V)
L = T - V
print('L')
display(L)
print('Bevægelsesligning (r)')
display(sp.Eq(L.diff(r.diff(t)).diff(t)-L.diff(r),0))
print('(expand og simplify)')
display(sp.Eq(L.diff(r.diff(t)).diff(t)-L.diff(r),0).expand().simplify())
print('Bevægelsesligning (theta)')
display(sp.Eq(L.diff(theta.diff(t)).diff(t)-L.diff(theta),0))
print('(expand og simplify)')
display(sp.Eq(L.diff(theta.diff(t)).diff(t)-L.diff(theta),0).expand().simplify())
````
